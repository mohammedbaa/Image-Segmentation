## Upsampling

```python
import torch 
import torch.nn.functional as F
```

```python
image=torch.tensor([
    [1,2],
    [3,4]
]).float().unsqueeze(0).unsqueeze(0)
```

## Nearest -neighbour 

```python
upsample_nn=F.interpolate(image , scale_factor=2,mode='nearest')
```

## Transposed Convolution 

```python
conv_transpose=torch.nn.ConvTranspose2d(1,1,kernel_size=3,stride=2,padding=1)
output_transpose=conv_transpose(image)
```

```python
print("Nearest neighbour Upsampling :" ,upsample_nn.squeeze())
print("Transpose Convolution output :" ,output_transpose.squeeze())
```
